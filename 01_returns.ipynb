```python
import numpy as np
import pandas as pd
import yfinance as yf
```

```python

tickers = ["NVDA","INTC", "AMD", "MU", "MRVL", "WDC", "AVGO", "SPY"]
prices = yf.download(tickers, period="5y", auto_adjust=True)["Close"]
prices.tail()
```

```python
log_returns = np.log(prices / prices.shift(1)).dropna()
log_returns.head()
```
